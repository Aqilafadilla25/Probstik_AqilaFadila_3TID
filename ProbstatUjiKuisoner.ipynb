```python
import re
import pandas as pd
import numpy as np
from scipy import stats
from google.colab import files
```

```python
up = files.upload()
nama_file = list(up.keys()) [0]
print("File terbaca:", nama_file)
```

```python
JUMLAH_KOLOM_DEMOGRAFI = 5
df = pd.read_excel(nama_file)
butir = df.iloc[:, JUMLAH_KOLOM_DEMOGRAFI:]

# Buang kolom yang bukan butir Likert (misalnya Durasi)
butir = butir.loc[:, butir.columns.str.contains(r" \[")]

print("Jumlah responden :", len(df))
print("Jumlah butir     :", butir.shape[1])
print("Data kosong      :", butir.isna().sum().sum())
```

```python
skala = {
    "Sangat Tidak Setuju (STS)": 1,
    "Tidak Setuju (TS)": 2,
    "Netral (N)": 3,
    "Setuju (S)": 4,
    "Sangat Setuju (SS)": 5,
}

num = butir.replace(skala).apply(pd.to_numeric, errors="coerce")

# Harus 0. Jika lebih dari 0, ada jawaban yang tidak terbaca.
print("Sel yang gagal dikonversi:", int(num.isna().sum().sum()))
num.head()
```

```python
dimensi = {}
for kolom in butir.columns:
  nama = re.match(r"^(.*?) \[", kolom).group(1)
  dimensi.setdefault(nama, []).append(kolom)

for nama, daftar in dimensi.items():
  print(f"{nama}: {len(daftar)} butir")
```

```python
n = len(num)
t = stats.t.ppf(0.975, n - 2)
r_tabel = t / np.sqrt((n - 2) + t**2)
print(f"n = {n} | r tabel (α=0,05; df={n-2}) = {r_tabel:.3f}")

hasil_validitas = []
for nama, daftar in dimensi.items():
  x = num [daftar]
  total = x.sum(axis=1)
  for i, kolom in enumerate(daftar, 1):
    r, p = stats.pearsonr(x[kolom], total)
    r_koreksi, _ = stats.pearsonr(x[kolom], total - x[kolom])
    hasil_validitas.append({
        "Dimensi": nama,
        "Butir": f"B{i}",
        "r hitung": round(r, 3),
        "Corrected r": round(r_koreksi, 3),
        "p-value": round(p,4),
        "Keterangan": "Valid" if r > r_tabel else "Tidak Valid",
    })

  tabel_validitas = pd.DataFrame(hasil_validitas)
  tabel_validitas
```

```python
def cronbach_alpha(x):
  k = x.shape[1]
  return k / (k - 1) * (1 - x.var(ddof=1).sum() / x.sum(axis=1).var(ddof=1))

def kategori(a):
  return("Sangat baik" if a >= 0.90 else "Baik" if a >= 0.80
         else "Reliabel" if a >= 0.70 else "Tidak reliabel")

hasil_reliabilitas = []
alpha_dihapus = []

for nama, daftar in dimensi.items():
  x = num[daftar]
  a = cronbach_alpha(x)
  hasil_reliabilitas.append({
      "Dimensi": nama,
      "Jumlah butir": len(daftar),
      "Cronbach's alpha": round(a,3),
      "Kategori": kategori(a),
  })

  for i, kolom in enumerate(daftar, 1):
    if len(daftar) > 2:
      alpha_dihapus.append({
          "Dimensi": nama,
          "Butir": f"B{i}",
          "Cronbach's alpha": round(cronbach_alpha(x.drop(columns=kolom)), 3)})

tabel_reliabilitas = pd.DataFrame(hasil_reliabilitas)
tabel_alpha_dihapus = pd.DataFrame(alpha_dihapus)
display(tabel_reliabilitas)
display(tabel_alpha_dihapus)
```

```python
gabung = tabel_validitas.merge(tabel_alpha_dihapus, on=["Dimensi", "Butir"],
how = "left")

with pd.ExcelWriter("hasil_uji_validitas_reliabilitas.xlsx") as w:
  gabung.to_excel(w, sheet_name="Validitas", index=False)
  tabel_reliabilitas.to_excel(w, sheet_name="Reliabilitas", index=False)

files.download("hasil_uji_validitas_reliabilitas.xlsx")
```

```python
# LATIHAN 1 - Interpretasi: butir tidak valid
print(f"r tabel = {r_tabel:.3f}")
tidak_valid = tabel_validitas[tabel_validitas["Keterangan"] != "Valid"]
if tidak_valid.empty:
    print("Tidak ada butir yang tidak valid.")
else:
    display(tidak_valid.assign(**{"r tabel": round(r_tabel, 3)}))

print("\n3 butir dengan r hitung paling rendah:")
display(tabel_validitas.sort_values("r hitung").head(3))
```

```python
#LATIHAN 2 - Perbandingan kriteria: r hitung vs corrected r
tv = tabel_validitas.copy()
tv["Selisih"] = (tv["r hitung"] - tv["Corrected r"]).round(3)

jml_r = int((tv["r hitung"] > r_tabel).sum())
jml_cr = int((tv["Corrected r"] >= 0.30).sum())
print(f"Butir valid (r hitung > r tabel {r_tabel:.3f}) : {jml_r} dari {len(tv)}")
print(f"Butir valid (corrected r >= 0,30)         : {jml_cr} dari {len(tv)}")

print("\nButir lolos r hitung tetapi corrected r < 0,30:")
display(tv[(tv["r hitung"] > r_tabel) & (tv["Corrected r"] < 0.30)])

print("\n5 selisih terbesar (r hitung - corrected r):")
display(tv.sort_values("Selisih", ascending=False).head(5))

print("\nSelisih terkecil:")
display(tv.sort_values("Selisih").head(3))

print("\nRata-rata selisih per dimensi:")
display(tv.groupby("Dimensi")["Selisih"].mean().round(3))
```

```python
# LATIHAN 3 - Perbaikan instrumen: buang butir, uji ulang
# Tidak ada butir tidak valid, jadi dibuang butir terlemah (r hitung terendah).
terlemah = tabel_validitas.sort_values("r hitung").iloc[0]
print("Butir terlemah:", terlemah["Dimensi"], terlemah["Butir"],
      "| r hitung =", terlemah["r hitung"], "| corrected r =", terlemah["Corrected r"])

def uji_ulang(nama_dim, butir_dibuang):
    daftar = dimensi[nama_dim]
    label = {kol: f"B{i}" for i, kol in enumerate(daftar, 1)}   # label B asli
    sisa = [kol for kol in daftar if label[kol] != butir_dibuang]
    x = num[sisa]
    total = x.sum(axis=1)
    baris = []
    for kol in sisa:
        r, p = stats.pearsonr(x[kol], total)
        rk, _ = stats.pearsonr(x[kol], total - x[kol])
        baris.append({"Butir": label[kol], "r hitung": round(r, 3),
                      "Corrected r": round(rk, 3), "p-value": round(p, 4),
                      "Keterangan": "Valid" if r > r_tabel else "Tidak Valid"})
    a_awal = cronbach_alpha(num[daftar])
    a_baru = cronbach_alpha(x)
    return a_awal, a_baru, pd.DataFrame(baris)

# Butir terlemah -> Aspek Psikologis B6
nama_dim = terlemah["Dimensi"]
a0, a1, hasil = uji_ulang(nama_dim, terlemah["Butir"])
print(f"\n{nama_dim}")
print(f"Alpha sebelum = {a0:.3f} ({kategori(a0)}) | sesudah {terlemah['Butir']} dibuang = {a1:.3f} ({kategori(a1)})")
display(hasil)

# Perbandingan alpha untuk semua butir yang perlu ditinjau
kol_alpha = [c for c in tabel_alpha_dihapus.columns if "lpha" in c][0]
tinjau = [("Penggunaan AI", "B4"), ("Aspek Psikologis", "B6"),
          ("Prokrastinasi", "B4"), ("Rsspon", "B4")]
rekap = []
for kunci, b in tinjau:
    nm = [d for d in dimensi if d.startswith(kunci)][0]
    a_awal = float(tabel_reliabilitas.loc[tabel_reliabilitas["Dimensi"] == nm, "Cronbach's alpha"].iloc[0])
    a_sesudah = float(tabel_alpha_dihapus[(tabel_alpha_dihapus["Dimensi"] == nm) &
                                          (tabel_alpha_dihapus["Butir"] == b)][kol_alpha].iloc[0])
    rekap.append({"Dimensi": nm, "Butir dibuang": b,
                  "Alpha sebelum": a_awal, "Alpha sesudah": a_sesudah})
print("\nRekap alpha sebelum vs sesudah:")
display(pd.DataFrame(rekap))
```

```python
# LATIHAN 4 - Pengaruh ukuran sampel (random_state = 1)
hasil_n = []
for m in [10, 15, 20, 25, 30, len(num)]:
    sub = num if m == len(num) else num.sample(m, random_state=1)
    t_m = stats.t.ppf(0.975, m - 2)
    r_m = t_m / np.sqrt((m - 2) + t_m**2)
    valid, gagal = 0, []
    for nama, daftar in dimensi.items():
        x = sub[daftar]
        total = x.sum(axis=1)
        for i, kolom in enumerate(daftar, 1):
            r, _ = stats.pearsonr(x[kolom], total)
            if r > r_m:
                valid += 1
            else:
                gagal.append(f"{' '.join(nama.split()[:2])} B{i}")
    hasil_n.append({"n": m, "r tabel": round(r_m, 3),
                    "Butir valid": f"{valid} dari {butir.shape[1]}",
                    "Butir tidak lolos": ", ".join(gagal) if gagal else "-"})
display(pd.DataFrame(hasil_n))
```

```python
# LATIHAN 5 - Refleksi: reliabel belum tentu valid (simulasi)
rng = np.random.default_rng(1)
waktu_asli = np.array([8.00, 10.00, 12.00, 15.00])        # jam sebenarnya
ukuran = np.array([waktu_asli - 5/60 + rng.normal(0, 0.005, 4) for _ in range(5)])

print("Hasil 5x pengukuran (jam):")
print(ukuran.round(3))
print("\nSimpangan baku antar pengukuran (kecil = konsisten/reliabel):",
      ukuran.std(axis=0, ddof=1).round(3))
print("Rata-rata selisih dari waktu asli (jam)  (jauh dari 0 = tidak akurat):",
      round((ukuran.mean(axis=0) - waktu_asli).mean(), 3))
```

```python
p_m = 0.05
sens = 0.90
false = 0.10
p_pos = sens * p_m + false * (1 - p_m)
p_bayes = sens * p_m / p_pos
print("P(positif):", p_pos)
print("P(malware | positif):", p_bayes)
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import binom

d = binom(10, 0.8)
print("P(X=8):", d.pmf(8))
print("Mean:", d.mean(), "Varians:", d.var())

x = np.arange(11)
plt.bar(x, d.pmf(x))
plt.title("Binomial")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import poisson

d = poisson(5)
print("P(X=3):", d.pmf(3), "Mean:", d.mean(), "Varians:", d.var())

x = np.arange(16)
plt.bar(x, d.pmf(x))
plt.title("Poisson")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm

d = norm(100, 15)
print("P(X<120):", d.cdf(120), "Mean:", d.mean(), "Varians:", d.var())

x = np.linspace(40, 160, 500)
plt.plot(x, d.pdf(x))
plt.axvline(120, ls="--")
plt.title("Normal")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import uniform

d = uniform(loc=2, scale=8)
print("P(4<=X<=7):", d.cdf(7) - d.cdf(4))
print("Mean:", d.mean(), "Varians:", d.var())

x = np.linspace(0, 12, 500)
plt.plot(x, d.pdf(x))
plt.title("Uniform")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import expon

d = expon(scale=5)
print("P(X<3):", d.cdf(3))
print("Mean:", d.mean(), "Varians:", d.var())

x = np.linspace(0, 20, 500)
plt.plot(x, d.pdf(x))
plt.title("Eksponensial")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(7)
ns = [100, 1000, 10000, 100000]
ps = []

for n in ns:
    d = np.random.randint(1, 7, n)
    p = np.mean(d == 6)
    ps.append(p)
    print(f"n={n:>6}: P empiris={p:.5f}")

plt.plot(ns, ps, marker="o", label="Simulasi")
plt.axhline(1 / 6, ls="--", label="Teori = 1/6")
plt.xscale("log")
plt.xlabel("Jumlah percobaan")
plt.ylabel("Frekuensi relatif angka 6")
plt.title("Konvergensi Simulasi Dadu")
plt.legend()
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import binom, geom, poisson

fig, ax = plt.subplots(1, 3, figsize=(12, 4))

x = np.arange(11)
ax[0].bar(x, binom.pmf(x, 10, 0.5))
ax[0].set_title("Binomial")

x = np.arange(16)
ax[1].bar(x, poisson.pmf(x, 5))
ax[1].set_title("Poisson")

x = np.arange(1, 16)
ax[2].bar(x, geom.pmf(x, 0.3))
ax[2].set_title("Geometrik")

plt.tight_layout()
plt.show()
```
